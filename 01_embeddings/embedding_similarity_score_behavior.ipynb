```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
# Fixed model for controlled experiment
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
sentence_pairs = [
    # Identical / Near-identical
    ("Machine learning is powerful", "Machine learning is powerful", "Identical"),
    ("Natural language processing is fun",
     "Natural language processing is fun", "Identical"),

    # Paraphrases
    ("I love studying machine learning",
     "I enjoy learning about machine learning", "Paraphrase"),
    ("She is building NLP systems",
     "She works on natural language processing projects", "Paraphrase"),

    # Topically related
    ("Transformers generate embeddings",
     "Sentence embeddings are used in semantic search", "Related"),
    ("Machine learning improves NLP tasks",
     "Natural language processing uses artificial intelligence", "Related"),

    # Unrelated
    ("I love machine learning",
     "The weather is very hot today", "Unrelated"),
    ("He plays cricket on weekends",
     "Sentence transformers create embeddings", "Unrelated")
]
```

```python
results = []

for sent_a, sent_b, category in sentence_pairs:
    emb_a = model.encode(sent_a)
    emb_b = model.encode(sent_b)

    score = cosine_similarity([emb_a], [emb_b])[0][0]

    results.append({
        "Sentence A": sent_a,
        "Sentence B": sent_b,
        "Category": category,
        "Cosine Similarity": round(score, 4)
    })

df = pd.DataFrame(results)
df
```

```python
df.groupby("Category")["Cosine Similarity"].agg(
    ["min", "max", "mean"]
).round(4)
```

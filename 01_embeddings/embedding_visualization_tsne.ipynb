```python
from sentence_transformers import SentenceTransformer
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
sentences = [
    "Machine learning improves data-driven decision making.",
    "Deep learning is a subset of machine learning.",
    "Neural networks are used in deep learning.",
    "Natural language processing deals with human language.",
    "Text classification is a common NLP task.",
    "Python is widely used for data science.",
    "Pandas and NumPy are popular Python libraries.",
    "Football is a popular sport worldwide.",
    "Cricket is widely played in India.",
    "Tennis is an individual sport."
]
```

```python
embeddings = model.encode(sentences)
embeddings.shape
```

```python
tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=5,
    learning_rate=200
)

reduced_embeddings = tsne.fit_transform(embeddings)
```

```python
plt.figure(figsize=(10, 7))

x = reduced_embeddings[:, 0]
y = reduced_embeddings[:, 1]

plt.scatter(x, y)

for i, sentence in enumerate(sentences):
    plt.annotate(sentence, (x[i], y[i]), fontsize=9)

plt.title("t-SNE Visualization of Sentence Embeddings")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.show()
```

- This notebook visualizes sentence embeddings using t-SNE to understand how semantically similar sentences are positioned closer together in vector space.
- High-dimensional sentence embeddings are reduced to two dimensions for visualization.
- Sentences with similar meanings tend to form clusters, demonstrating how embedding models capture semantic relationships.

```python
def evaluate_answer(question, answer):
    score = 0
    reason = []

    # Check length
    if len(answer.split()) > 3:
        score += 1
    else:
        reason.append("Answer too short")

    # Check relevance (simple keyword match)
    if any(word.lower() in answer.lower() for word in question.split()):
        score += 1
    else:
        reason.append("Answer not relevant to question")

    # Check for nonsense words
    if "banana" in answer.lower():
        reason.append("Answer contains irrelevant content")

    # Final evaluation
    if score == 2 and not reason:
        return "GOOD ✅", "Clear and relevant answer"
    else:
        return "BAD ❌", ", ".join(reason)
```

```python
q1 = "What is recursion?"
a1 = "Recursion is when a function calls itself."

q2 = "What is recursion?"
a2 = "Recursion is a banana concept."

q3 = "Explain machine learning"
a3 = "Machine learning is a method where systems learn from data."

for q, a in [(q1, a1), (q2, a2), (q3, a3)]:
    result, reason = evaluate_answer(q, a)
    print("\n🧠 Question:", q)
    print("💬 Answer:", a)
    print("📊 Evaluation:", result)
    print("📝 Reason:", reason)
```

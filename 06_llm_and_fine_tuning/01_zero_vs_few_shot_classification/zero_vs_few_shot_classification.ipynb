```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")

```

```python
text = "The government passed a new education reform bill in parliament."

labels = ["Sports", "Politics", "Technology"]

result = classifier(text, labels)

print("ZERO-SHOT RESULT:")
print(result)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
prompt = """
Classify the text into Sports, Politics, or Technology.

Examples:
India won the cricket world cup → Sports
New AI chip released by Nvidia → Technology
Election campaign started in Delhi → Politics

Text: Government passed a new bill on education →
"""
```

```python
inputs = tokenizer(prompt, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=10
)

result = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("FEW-SHOT RESULT:")
print(result)
```

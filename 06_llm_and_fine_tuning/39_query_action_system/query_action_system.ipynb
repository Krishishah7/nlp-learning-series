```python
import re
```

```python
def calculator(query):
    expr = "".join([c for c in query if c in "0123456789+-*/"])
    try:
        return str(eval(expr))
    except:
        return "Invalid calculation"

def explain(query):
    return f"{query} is a concept that can be explained step-by-step in simple terms."

def knowledge(query):
    return f"{query} refers to a well-known fact or entity. (Can be extended with real data)"
```

```python
def decide_action(query):
    query = query.lower()

    if re.search(r"\d+[\+\-\*/]\d+", query):
        return "calculator"

    elif any(word in query for word in ["explain", "what is", "define"]):
        return "explain"

    elif any(word in query for word in ["who is", "when", "where"]):
        return "knowledge"

    else:
        return "unknown"
```

```python
def agent_system(query):
    action = decide_action(query)

    print("\n🧠 Query:", query)
    print("🤖 Agent Decision:", action)

    if action == "calculator":
        result = calculator(query)

    elif action == "explain":
        result = explain(query)

    elif action == "knowledge":
        result = knowledge(query)

    else:
        result = "Agent cannot handle this query."

    print("✅ Result:", result)
```

```python
agent_system("Calculate 25*4")
agent_system("Explain recursion")
agent_system("Who is Elon Musk?")
agent_system("Tell me a joke")
```

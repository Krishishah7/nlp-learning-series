```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
prompt = "Explain why exercise is good for health in two sentences."
```

```python
responses = []

for _ in range(5):
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=60,
        temperature=0.9,
        do_sample=True
    )

    text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    responses.append(text)

print("MODEL RESPONSES:\n")

for i, r in enumerate(responses, 1):
    print(f"Response {i}: {r}\n")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(responses)

similarity_matrix = cosine_similarity(X)

print("SIMILARITY MATRIX:\n")
print(similarity_matrix)
```

```python
import numpy as np

avg_similarity = (np.sum(similarity_matrix) - len(responses)) / (len(responses)*(len(responses)-1))

print("AVERAGE RESPONSE SIMILARITY:")
print(avg_similarity)
```

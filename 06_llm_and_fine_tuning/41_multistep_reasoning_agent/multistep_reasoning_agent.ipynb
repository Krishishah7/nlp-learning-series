```python
import re
```

```python
def calculator(expr):
    try:
        return str(eval(expr))
    except:
        return "Invalid calculation"

def explain(text, is_math=False):
    if is_math:
        return f"The result {text} is obtained by performing the calculation step-by-step."
    else:
        return f"{text} is a concept that can be understood by breaking it into smaller steps."
```

```python
def detect_steps(query):
    query = query.lower()
    steps = []

    if re.search(r"\d+[\+\-\*/]\d+", query):
        steps.append("math")

    if "explain" in query:
        steps.append("explain")

    return steps
```

```python
def multistep_agent(query):
    steps = detect_steps(query)

    print("\n🧠 Query:", query)
    print("🪜 Steps:", steps)

    result = None

    for step in steps:
        if step == "math":
            expr = "".join([c for c in query if c in "0123456789+-*/"])
            result = calculator(expr)
            print("🔢 Math Result:", result)

        elif step == "explain":
          if result:
              explanation = explain(result, is_math=True)
          else:
              topic = query.replace("explain", "").strip()
              explanation = explain(topic, is_math=False)

          print("📘 Explanation:", explanation)

    if not steps:
        print("⚠️ No steps detected.")
```

```python
multistep_agent("Calculate 5*10 and explain the result")
multistep_agent("Explain recursion")
multistep_agent("What is 8+2 and explain")
multistep_agent("Tell me a joke")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
prompt_bad = "Tell me about AI."

inputs = tokenizer(prompt_bad, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=60)

result_bad = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("POOR PROMPT OUTPUT:\n")
print(result_bad)
```

```python
prompt_good = """
Explain Artificial Intelligence in simple terms for a school student.
Limit your answer to 3 short sentences.
"""

inputs = tokenizer(prompt_good, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=60)

result_good = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("GOOD PROMPT OUTPUT:\n")
print(result_good)
```

```python
prompt_role = """
You are a helpful science teacher.

Explain Artificial Intelligence to a 10-year-old child using a simple example.
"""

inputs = tokenizer(prompt_role, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=60)

result_role = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("ROLE PROMPT OUTPUT:\n")
print(result_role)
```

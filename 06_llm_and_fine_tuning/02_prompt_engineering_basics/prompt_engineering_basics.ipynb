```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-large"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
prompt_bad = "This movie was amazing!"

inputs = tokenizer(prompt_bad, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=10)

print("POOR PROMPT OUTPUT:")
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
prompt_good = """
Classify the sentiment of the following text as Positive, Negative, or Neutral.

Text: This movie was amazing!
"""

inputs = tokenizer(prompt_good, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=10)

print("GOOD PROMPT OUTPUT:")
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
prompt_role = """
You are a sentiment analysis expert.

Classify the sentiment as Positive, Negative, or Neutral.

Text: This movie was amazing!
"""

inputs = tokenizer(prompt_role, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=10)

print("ROLE PROMPT OUTPUT:")
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

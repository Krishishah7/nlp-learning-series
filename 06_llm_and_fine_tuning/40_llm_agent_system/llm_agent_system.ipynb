```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
def generate(prompt):
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=20,
        temperature=0.2
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
def decide_action_llm(query):
    prompt = f"""
You are an intelligent AI agent.

Your task is to choose the correct action for the query.

Available actions:
- calculator → for math calculations
- explain → for explaining concepts
- knowledge → for facts about people, places, or things

Return ONLY one word: calculator, explain, or knowledge.

Query: {query}
Action:
"""

    result = generate(prompt).strip().lower()

    if "calculator" in result:
        return "calculator"
    elif "explain" in result:
        return "explain"
    elif "knowledge" in result:
        return "knowledge"
    else:
        return "unknown"
```

```python
def calculator(query):
    expr = "".join([c for c in query if c in "0123456789+-*/"])
    try:
        return str(eval(expr))
    except:
        return "Invalid calculation"

def explain(query):
    return f"{query} is a concept explained simply."

def knowledge(query):
    return f"{query} refers to a known fact."
```

```python
def llm_agent(query):
    action = decide_action_llm(query)

    print("\n🧠 Query:", query)
    print("🤖 LLM Decision:", action)

    if action == "calculator":
        result = calculator(query)
    elif action == "explain":
        result = explain(query)
    elif action == "knowledge":
        result = knowledge(query)
    else:
        result = "Agent could not decide action."

    print("✅ Result:", result)
```

```python
llm_agent("Calculate 25*4")
llm_agent("Explain recursion")
llm_agent("Who is Elon Musk?")
llm_agent("Tell me a joke")
```

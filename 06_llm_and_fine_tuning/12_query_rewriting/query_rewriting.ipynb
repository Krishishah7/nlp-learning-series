```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
user_query = "capital france?"
```

```python
rewrite_prompt = f"""
Rewrite the following question into a clear and complete sentence:

{user_query}
"""

inputs = tokenizer(rewrite_prompt, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=40)

rewritten_query = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("ORIGINAL QUERY:")
print(user_query)

print("\nREWRITTEN QUERY:")
print(rewritten_query)
```

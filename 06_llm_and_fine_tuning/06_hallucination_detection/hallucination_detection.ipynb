```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
questions = [
    "What is the capital of France?",
    "Who wrote Romeo and Juliet?",
    "What planet is known as the Red Planet?"
]

ground_truth = [
    "Paris",
    "William Shakespeare",
    "Mars"
]
```

```python
answers = []

for q in questions:
    prompt = f"Answer briefly: {q}"

    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=20
    )

    text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    answers.append(text)

print("MODEL ANSWERS:\n")

for q, a in zip(questions, answers):
    print(f"Q: {q}")
    print(f"A: {a}\n")
```

```python
print("HALLUCINATION CHECK:\n")

for q, pred, truth in zip(questions, answers, ground_truth):
    if truth.lower() in pred.lower():
        status = "Correct"
    else:
        status = "Hallucinated / Incorrect"

    print(f"Q: {q}")
    print(f"Model Answer: {pred}")
    print(f"Ground Truth: {truth}")
    print(f"Status: {status}\n")
```

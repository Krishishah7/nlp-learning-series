```python
def optimize_prompt(prompt):
    prompt_lower = prompt.lower()

    # If explanation type
    if "explain" in prompt_lower or "what is" in prompt_lower:
        return f"{prompt.strip()} in simple terms with an example and key points."

    # If "tell me about"
    elif "tell me about" in prompt_lower:
        topic = prompt_lower.replace("tell me about", "").strip()
        return f"Explain {topic} with its features, uses, and a simple example."

    # If very short prompt
    elif len(prompt.split()) <= 2:
        return f"Explain {prompt.strip()} in detail with examples and key points."

    # Default improvement
    else:
        return f"{prompt.strip()} with clear explanation, examples, and structured points."
```

```python
prompts = [
    "Explain AI",
    "Tell me about Python",
    "Recursion",
    "Machine learning basics"
]

for p in prompts:
    print("\n🧠 Original:", p)
    print("✨ Improved:", optimize_prompt(p))
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
question = "If a box has 5 apples and you add 3 more apples, how many apples are there in total?"
```

```python
prompt_direct = f"""
Answer the question:

{question}
"""

inputs = tokenizer(prompt_direct, return_tensors="pt")

outputs = model.generate(**inputs, max_new_tokens=20)

print("DIRECT PROMPT OUTPUT:")
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
prompt_cot = f"""
Answer the question step by step.

{question}
"""

inputs = tokenizer(prompt_cot, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=60,
    temperature=0.7,
    do_sample=True
)

print("CHAIN-OF-THOUGHT OUTPUT:")
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

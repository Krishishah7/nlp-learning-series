```python
import numpy as np
import pandas as pd
```

```python
space_docs = [
    "Elon Musk founded SpaceX.",
    "SpaceX works on rockets and space exploration."
]

auto_docs = [
    "Tesla builds electric cars.",
    "Elon Musk is CEO of Tesla."
]

space_df = pd.DataFrame({"text": space_docs})
auto_df = pd.DataFrame({"text": auto_docs})
```

```python
def classify_query(query):
    q = query.lower()

    if any(word in q for word in ["rocket", "space", "spacex"]):
        return "space"
    elif any(word in q for word in ["car", "tesla", "electric"]):
        return "automobile"
    else:
        return "unknown"
```

```python
def retrieve(query, df):
    scores = []

    for text in df["text"]:
        score = sum(word in text.lower() for word in query.lower().split())
        scores.append(score)

    scores = np.array(scores)
    top_idx = np.argmax(scores)

    return df.iloc[top_idx]["text"], scores[top_idx]
```

```python
def extract_answer(context):
    if "founded" in context.lower():
        return "Elon Musk"
    elif "spacex" in context.lower():
        return "SpaceX"
    elif "tesla" in context.lower():
        return "Tesla"
    else:
        return "Answer not found"
```

```python
def routed_rag(query):
    domain = classify_query(query)

    if domain == "space":
        context, score = retrieve(query, space_df)

    elif domain == "automobile":
        context, score = retrieve(query, auto_df)

    else:
        print("\nUnknown Domain 🚫")
        print("Response: This query is outside supported domains.")
        return

    answer = extract_answer(context)

    print("\nQuery:", query)
    print("Domain:", domain)
    print("Context:", context)
    print("Answer:", answer)
```

```python
routed_rag("Which company works on rockets?")
routed_rag("Which company builds electric cars?")
routed_rag("What is the weather today?")
```

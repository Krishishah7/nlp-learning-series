```python
import re
```

```python
def detect_query_type(query):
    query = query.lower()

    # Math detection
    if re.search(r"\d+[\+\-\*/]\d+", query):
        return "math"

    # Explanation detection
    elif any(word in query for word in ["explain", "what is", "define"]):
        return "education"

    # Knowledge detection
    elif any(word in query for word in ["who is", "when", "where"]):
        return "knowledge"

    else:
        return "unknown"
```

```python
def calculator(query):
    expr = "".join([c for c in query if c in "0123456789+-*/"])
    try:
        return str(eval(expr))
    except:
        return "Invalid expression"
```

```python
def explain(query):
    return f"This is a simple explanation for: {query}"
```

```python
def knowledge(query):
    return f"This is factual information about: {query}"
```

```python
def routing_system(query):
    qtype = detect_query_type(query)

    print("\n🧠 Query:", query)
    print("🔀 Route:", qtype)

    if qtype == "math":
        result = calculator(query)

    elif qtype == "education":
        result = explain(query)

    elif qtype == "knowledge":
        result = knowledge(query)

    else:
        result = "Sorry, I cannot handle this query."

    print("✅ Response:", result)
```

```python
routing_system("Explain recursion")
routing_system("Calculate 25*4")
routing_system("Who is Elon Musk?")
routing_system("Tell me a joke")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-large"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
def generate(prompt):
    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        **inputs,
        max_new_tokens=20,
        temperature=0.2
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
def classify_query(query):
    categories = ["Education", "Math", "Knowledge"]

    examples = [
        ("Explain recursion", "Education"),
        ("Calculate 10 + 5", "Math"),
        ("Who is Albert Einstein?", "Knowledge")
    ]

    example_text = ""
    for q, c in examples:
        example_text += f"Query: {q}\nCategory: {c}\n\n"

    prompt = f"""
Classify the query into one of these categories:
{", ".join(categories)}

Here are some examples:
{example_text}

Now classify:

Query: {query}
Category:
"""

    result = generate(prompt)
    return result.strip()
```

```python
def query_classifier_app(query):
    category = classify_query(query)

    print("\n🧠 Query:", query)
    print("📂 Category:", category)
```

```python
query_classifier_app("Explain recursion")
query_classifier_app("25 * 4")
query_classifier_app("Who is Elon Musk?")
query_classifier_app("What is machine learning?")
```

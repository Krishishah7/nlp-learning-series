```python
memory = []

def update_memory(query, response):
    memory.append({
        "query": query,
        "response": response
    })
```

```python
def get_context():
    for item in reversed(memory):
        if "explain" in item["query"].lower() or "what is" in item["query"].lower():
            return item["query"]
    return ""
```

```python
def explain(query):
    return f"Explanation: {query} is explained simply."

def example(query):
    return f"Example: Here is an example related to {query}."
```

```python
def memory_agent(query):
    context = get_context()

    print("\n🧠 Query:", query)
    print("🧾 Context:", context)

    if "example" in query.lower() and context:
        result = example(context)
    else:
        result = explain(query)

    update_memory(query, result)

    print("✅ Response:", result)
```

```python
memory_agent("What is recursion?")
memory_agent("Give example")
memory_agent("Explain machine learning")
memory_agent("Give example")
```

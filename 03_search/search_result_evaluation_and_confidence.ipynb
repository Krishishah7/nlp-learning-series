# Search Result Evaluation and Confidence Analysis

## Objective
This notebook evaluates semantic search results beyond ranking
by analyzing similarity score gaps to estimate confidence
in the top retrieved result.

The goal is to understand when search results are reliable
and when they are ambiguous.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
query = "learn machine learning basics"

documents = [
    "Machine learning tutorials for beginners",
    "Introduction to deep learning concepts",
    "Natural language processing using transformers",
    "Python programming fundamentals",
    "Top travel destinations in summer",
    "Latest football match highlights"
]
```

```python
query_emb = model.encode(query)
doc_embs = model.encode(documents)

scores = cosine_similarity([query_emb], doc_embs)[0]

df = pd.DataFrame({
    "Document": documents,
    "Similarity Score": np.round(scores, 4)
}).sort_values(by="Similarity Score", ascending=False)

df
```

```python
df["Score Gap From Top"] = df.iloc[0]["Similarity Score"] - df["Similarity Score"]
df
```

## Observations

- A large score gap between Rank-1 and Rank-2 indicates high confidence
- Small score gaps suggest ambiguity between results
- Confidence depends on relative scores, not absolute values

## Key Insight
Search systems should evaluate confidence using score gaps,
not just return the top-ranked document blindly.

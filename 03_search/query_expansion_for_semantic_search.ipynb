# Query Expansion for Semantic Search

## Objective
This notebook demonstrates how query expansion can improve
semantic search results by generating related queries and
combining their retrieval outputs.

The goal is to improve recall and intent coverage
without changing the underlying embedding model.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
original_query = "learn machine learning"

expanded_queries = [
    original_query,
    "machine learning tutorials for beginners",
    "introduction to machine learning",
    "basic machine learning concepts"
]
```

```python
documents = [
    "Machine learning tutorials for beginners",
    "Deep learning and neural networks explained",
    "Introduction to artificial intelligence",
    "Python programming fundamentals",
    "Natural language processing using transformers",
    "Football match highlights",
    "Top travel destinations"
]
```

```python
doc_embeddings = model.encode(documents)
```

```python
rows = []

for query in expanded_queries:
    query_emb = model.encode(query)
    scores = cosine_similarity([query_emb], doc_embeddings)[0]

    for doc, score in zip(documents, scores):
        rows.append({
            "Query Used": query,
            "Document": doc,
            "Similarity Score": score
        })

df = pd.DataFrame(rows)
df
```

```python
final_scores = (
    df.groupby("Document")["Similarity Score"]
      .max()
      .reset_index()
      .sort_values(by="Similarity Score", ascending=False)
)

final_scores
```

## Observations

- Expanded queries retrieve documents missed by the original query
- Taking the maximum similarity score improves recall
- Query expansion reduces dependence on exact phrasing

## Key Insight
Query expansion improves semantic search quality by broadening
intent coverage while keeping the same embedding model.

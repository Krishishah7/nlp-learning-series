# Query Type Analysis in Semantic Search

## Objective
This notebook analyzes how semantic search behaves across
different types of queries to understand retrieval robustness.

The goal is to observe how similarity scores and rankings
change depending on query intent and phrasing.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "Machine learning tutorials for beginners",
    "Deep learning and neural networks explained",
    "Natural language processing using transformers",
    "Python programming fundamentals",
    "Football match highlights",
    "Top travel destinations"
]
```

```python
queries = {
    "Informational": "learn machine learning basics",
    "Conceptual": "what is deep learning",
    "Programming": "python machine learning tutorial",
    "Unrelated": "best football players"
}
```

```python
doc_embs = model.encode(documents)

results = []

for query_type, query in queries.items():

    query_emb = model.encode(query)
    scores = cosine_similarity([query_emb], doc_embs)[0]

    for doc, score in zip(documents, scores):
        results.append({
            "Query Type": query_type,
            "Query": query,
            "Document": doc,
            "Similarity Score": round(score, 4)
        })

df = pd.DataFrame(results)
df
```

```python
df_sorted = df.sort_values(
    ["Query Type", "Similarity Score"],
    ascending=[True, False]
)

df_sorted
```

## Observations

- Informational queries retrieve tutorial-style documents effectively
- Conceptual queries retrieve explanation-style documents
- Programming queries prioritize coding-related documents
- Unrelated queries correctly retrieve non-NLP content

## Key Insight
Semantic search behavior depends strongly on query intent,
highlighting the importance of query understanding in
retrieval systems.

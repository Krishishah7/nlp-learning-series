# End-to-End Semantic Search Pipeline Demo

## Objective
This notebook demonstrates a complete semantic search pipeline
including embedding generation, similarity computation,
document ranking, and confidence evaluation.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "Machine learning tutorials for beginners",
    "Deep learning and neural networks explained",
    "Natural language processing with transformers",
    "Python programming fundamentals",
    "Football match highlights",
    "Top travel destinations"
]
```

```python
query = "learn machine learning basics"
```

```python
query_emb = model.encode(query)
doc_embs = model.encode(documents)
```

```python
scores = cosine_similarity([query_emb], doc_embs)[0]
```

```python
df = pd.DataFrame({
    "Document": documents,
    "Similarity Score": np.round(scores,4)
}).sort_values(by="Similarity Score", ascending=False)

df
```

```python
top_k = 3
top_results = df.head(top_k)

top_results
```

```python
top_score = df.iloc[0]["Similarity Score"]
second_score = df.iloc[1]["Similarity Score"]

score_gap = top_score - second_score

confidence = "High" if score_gap > 0.15 else "Low"

print("Confidence:", confidence)
```

## Observations

The pipeline demonstrates how semantic search retrieves and ranks
documents based on meaning rather than keyword matching.

Confidence estimation helps determine whether the system is
certain about the top result.

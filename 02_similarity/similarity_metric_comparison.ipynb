# Similarity Metric Comparison

## Objective
This notebook compares different similarity and distance metrics
used in NLP to understand how they behave when applied to
sentence embeddings.

The goal is to justify why cosine similarity is commonly preferred
for semantic similarity tasks.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity, euclidean_distances
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
pairs = [
    ("I love machine learning", "I enjoy studying machine learning"),
    ("Transformers generate embeddings", "Sentence embeddings help semantic search"),
    ("Python is a programming language", "Python is a dangerous snake"),
    ("I love machine learning", "The weather is sunny today")
]
```

```python
sentences = list(set([s for pair in pairs for s in pair]))
embeddings = {s: model.encode(s) for s in sentences}
```

```python
rows = []

for a, b in pairs:
    emb_a = embeddings[a]
    emb_b = embeddings[b]

    rows.append({
        "Sentence A": a,
        "Sentence B": b,
        "Cosine Similarity": round(cosine_similarity([emb_a], [emb_b])[0][0], 4),
        "Euclidean Distance": round(euclidean_distances([emb_a], [emb_b])[0][0], 4),
        "Dot Product": round(np.dot(emb_a, emb_b), 4)
    })

df = pd.DataFrame(rows)
df
```

## Observations

- Cosine similarity is scale-invariant and focuses on direction
- Euclidean distance is affected by vector magnitude
- Dot product grows with vector size and is harder to interpret

## Key Insight
Cosine similarity aligns best with how sentence embeddings are trained
and interpreted, which is why it is widely used in semantic search
and retrieval systems.

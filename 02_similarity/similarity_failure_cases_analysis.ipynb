# Similarity Failure Case Analysis

## Objective
This notebook explores failure cases in semantic similarity scoring
using sentence embeddings.

The goal is to understand:
- When high similarity scores are misleading
- When low similarity scores hide meaningful relationships
- Why semantic similarity must be used with care in real systems

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
failure_cases = [
    # False positives (high score, wrong meaning)
    ("Python is a programming language", "Python is a dangerous snake", "False Positive"),
    ("Apple released a new product", "Apple is a healthy fruit", "False Positive"),

    # False negatives (low score, related meaning)
    ("He was fired from his job", "He lost his employment", "False Negative"),
    ("She passed away yesterday", "She is no longer alive", "False Negative"),

    # Ambiguous similarity
    ("The bank approved the loan", "The river bank was flooded", "Ambiguous"),
    ("He booked a flight", "He traveled to another country", "Ambiguous")
]
```

```python
rows = []

for a, b, case_type in failure_cases:
    emb_a = model.encode(a)
    emb_b = model.encode(b)
    score = cosine_similarity([emb_a], [emb_b])[0][0]

    rows.append({
        "Sentence A": a,
        "Sentence B": b,
        "Case Type": case_type,
        "Cosine Similarity": round(score, 4)
    })

df = pd.DataFrame(rows)
df
```

## Observations

- Some sentence pairs receive high similarity scores despite having different meanings
- Some semantically equivalent sentences receive lower-than-expected scores
- Ambiguity and polysemy are major sources of failure

## Key Insight
Sentence embeddings capture general semantic proximity,
not exact meaning or intent.

This is why ranking, context expansion, and human-in-the-loop
designs are often required in real NLP systems.

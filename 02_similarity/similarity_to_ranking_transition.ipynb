# From Similarity Thresholds to Ranking-Based Retrieval

## Objective
This notebook demonstrates why ranking-based retrieval (Top-K)
is more reliable than hard similarity thresholds when using
sentence embeddings.

The goal is to transition from similarity analysis
to real-world retrieval system design.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
query = "I want to learn machine learning"

documents = [
    "Machine learning tutorials for beginners",
    "Deep learning and neural networks explained",
    "Natural language processing with transformers",
    "Best places to visit in summer",
    "Football match highlights from yesterday",
    "Python programming basics"
]
```

```python
query_emb = model.encode(query)
doc_embs = model.encode(documents)

scores = cosine_similarity([query_emb], doc_embs)[0]

df = pd.DataFrame({
    "Document": documents,
    "Similarity Score": np.round(scores, 4)
})

df.sort_values(by="Similarity Score", ascending=False)
```

```python
threshold = 0.6

threshold_results = df[df["Similarity Score"] >= threshold]
threshold_results
```

```python
top_k = 3

top_k_results = df.sort_values(
    by="Similarity Score", ascending=False
).head(top_k)

top_k_results
```

## Observations

- Threshold filtering may return too few or zero results
- Slight score variations can remove relevant documents
- Top-K ranking always returns the most relevant candidates

## Key Insight
Ranking-based retrieval is more robust than threshold-based
filtering for semantic similarity tasks.

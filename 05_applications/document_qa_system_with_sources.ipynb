# Document QA System with Source Attribution

## Objective

Build a question-answering system that retrieves relevant
documents and returns answers along with supporting sources.

```python
!pip install faiss-cpu
```

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "BERT is a transformer-based model for language understanding.",
    "Transformers use self-attention to process text.",
    "Machine learning allows systems to learn from data.",
    "Deep learning uses neural networks with many layers.",
    "Python is widely used for AI and data science."
]
```

```python
doc_embeddings = model.encode(documents).astype("float32")

dimension = doc_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(doc_embeddings)
```

```python
query = "How does BERT process text?"
```

```python
query_embedding = model.encode([query]).astype("float32")

k = 3

distances, indices = index.search(query_embedding, k)

retrieved_docs = [documents[i] for i in indices[0]]
scores = distances[0]
```

```python
answer = retrieved_docs[0]
```

```python
print("Question:", query)
print("\nAnswer:", answer)

print("\nSources:")

for doc, score in zip(retrieved_docs, scores):
    print(f"- {doc} (distance: {score:.4f})")
```

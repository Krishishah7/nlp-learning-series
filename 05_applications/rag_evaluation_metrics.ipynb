# RAG Evaluation Metrics

## Objective

Evaluate retrieval and answer quality of a RAG system
using standard metrics.

```python
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
query = "How does BERT work?"

relevant_docs = [
    "BERT is a transformer-based model for language understanding."
]

retrieved_docs = [
    "Transformers use self-attention to process text.",
    "BERT is a transformer-based model for language understanding.",
    "Machine learning models learn patterns from data."
]
```

```python
def precision_at_k(retrieved, relevant, k):

    retrieved_k = retrieved[:k]

    hits = sum(doc in relevant for doc in retrieved_k)

    return hits / k

print("Precision@3:", precision_at_k(retrieved_docs, relevant_docs, 3))
```

```python
def recall_at_k(retrieved, relevant, k):

    retrieved_k = retrieved[:k]

    hits = sum(doc in relevant for doc in retrieved_k)

    return hits / len(relevant)

print("Recall@3:", recall_at_k(retrieved_docs, relevant_docs, 3))
```

```python
def reciprocal_rank(retrieved, relevant):

    for i, doc in enumerate(retrieved, 1):
        if doc in relevant:
            return 1 / i

    return 0

print("MRR:", reciprocal_rank(retrieved_docs, relevant_docs))
```

```python
generated_answer = "BERT uses transformers and self-attention to understand text."

reference_answer = "BERT is a transformer-based model that uses self-attention."

emb1 = model.encode(generated_answer)
emb2 = model.encode(reference_answer)

sim = cosine_similarity([emb1], [emb2])[0][0]

print("Answer Similarity:", sim)
```

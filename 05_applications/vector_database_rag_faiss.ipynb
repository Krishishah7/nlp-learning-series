# Vector Database RAG using FAISS

## Objective

Build a scalable RAG retrieval system using FAISS to store
and search embeddings efficiently.

```python
import faiss
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "BERT is a transformer-based model for language understanding.",
    "Transformers use self-attention to process text.",
    "Machine learning allows systems to learn patterns from data.",
    "Python is widely used for AI and data science.",
    "Deep learning uses neural networks with many layers."
]
```

```python
doc_embeddings = model.encode(documents)

doc_embeddings = np.array(doc_embeddings).astype("float32")
```

```python
dimension = doc_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(doc_embeddings)
```

```python
query = "How does BERT work?"

query_embedding = model.encode([query]).astype("float32")
```

```python
k = 3  # Top-K results

distances, indices = index.search(query_embedding, k)
```

```python
retrieved_docs = [documents[i] for i in indices[0]]

for i, doc in enumerate(retrieved_docs, 1):
    print(f"{i}. {doc}")
```

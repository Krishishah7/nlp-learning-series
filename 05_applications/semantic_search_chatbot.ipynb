# Semantic Search Chatbot

## Objective

Build a simple chatbot that answers user questions by retrieving
the most relevant document using semantic search.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "Machine learning tutorials help beginners understand core ML concepts.",
    "Deep learning is a subset of machine learning using neural networks.",
    "Natural language processing focuses on understanding human language.",
    "Python is widely used for machine learning and data science.",
    "Transformers are powerful models used for NLP tasks."
]
```

```python
doc_embeddings = model.encode(documents)
```

```python
query = "How do I start learning machine learning?"

query_embedding = model.encode(query)
```

```python
scores = cosine_similarity([query_embedding], doc_embeddings)[0]

best_index = np.argmax(scores)
best_answer = documents[best_index]
```

```python
print("User:", query)
print("Bot:", best_answer)
```

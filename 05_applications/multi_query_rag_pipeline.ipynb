# Multi-Query Retrieval for RAG

## Objective

Improve RAG retrieval quality by generating multiple
query variations and combining results.

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "BERT is a transformer-based model for language understanding.",
    "Transformers use self-attention to process text.",
    "Self-attention helps models focus on important words.",
    "Machine learning models learn patterns from data.",
    "Deep learning uses neural networks with many layers."
]
```

```python
doc_embeddings = model.encode(documents).astype("float32")

dimension = doc_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(doc_embeddings)
```

```python
query = "How does BERT work?"
```

```python
queries = [
    query,
    "What is BERT architecture?",
    "How do transformers process text?",
    "Explain self-attention in BERT"
]
```

```python
retrieved_indices = set()

for q in queries:

    q_emb = model.encode([q]).astype("float32")

    _, inds = index.search(q_emb, k=2)

    retrieved_indices.update(inds[0])
```

```python
results = [documents[i] for i in retrieved_indices]

for doc in results:
    print("-", doc)
```

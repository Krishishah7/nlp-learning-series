```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "Machine learning enables computers to learn from data.",
    "Natural language processing helps machines understand human language.",
    "Deep learning is a subset of machine learning using neural networks.",
    "Python is widely used in data science and artificial intelligence.",
    "Search engines retrieve relevant documents based on user queries."
]
```

```python
document_embeddings = model.encode(documents)
document_embeddings.shape
```

```python
query = "How do machines understand human language?"
```

```python
query_embedding = model.encode([query])
```

```python
similarity_scores = cosine_similarity(query_embedding, document_embeddings)[0]
similarity_scores
```

```python
top_k = 3

top_indices = np.argsort(similarity_scores)[::-1][:top_k]
top_indices
```

```python
results = pd.DataFrame({
    "Document": [documents[i] for i in top_indices],
    "Similarity Score": [similarity_scores[i] for i in top_indices]
})

results
```

```python
print("User Query:")
print(query)
print("\nTop Matching Documents:\n")

for i in top_indices:
    print("Score:", round(similarity_scores[i], 3))
    print("Document:", documents[i])
    print("-" * 70)
```

- This notebook implements a mini semantic document search application.
- Documents and user queries are converted into sentence embeddings using a transformer-based model.
- Cosine similarity is used to compare the query with all documents, and the top-k most relevant documents are retrieved based on semantic meaning rather than exact keyword matching.

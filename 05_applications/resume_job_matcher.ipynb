```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
resume_text = """
Experienced data scientist skilled in machine learning,
natural language processing, Python, and deep learning.
Worked on building semantic search and text classification systems.
"""
```

```python
job_description = """
We are looking for a machine learning engineer with experience
in NLP, deep learning, and Python to build intelligent search systems.
"""
```

```python
resume_embedding = model.encode([resume_text])
job_embedding = model.encode([job_description])
```

```python
similarity_score = cosine_similarity(resume_embedding, job_embedding)[0][0]

print("Resume-Job Match Score:", round(similarity_score, 3))
```

```python
if similarity_score > 0.75:
    print("Strong Match ✅")
elif similarity_score > 0.5:
    print("Moderate Match ⚠️")
else:
    print("Low Match ❌")
```

- This notebook implements a semantic resume-to-job description matching system.
- Both the resume and job description are converted into embeddings using a sentence transformer model.
- Cosine similarity is computed to measure how well the resume matches the job requirements.
- This approach can be extended to real-world recruitment systems.

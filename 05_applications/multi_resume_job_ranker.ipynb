```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
job_description = """
We are looking for a machine learning engineer with experience in
NLP, deep learning, Python, and building intelligent search systems.
"""
```

```python
resumes = [
    """
    Data scientist skilled in machine learning, NLP, and Python.
    Built semantic search and deep learning systems.
    """,

    """
    Frontend developer experienced in HTML, CSS, and JavaScript.
    Focused on UI/UX design and responsive web applications.
    """,

    """
    Machine learning engineer with experience in deep learning,
    natural language processing, and model deployment.
    """,

    """
    Software engineer with experience in databases and backend systems.
    Worked mainly with Java and MySQL.
    """
]
```

```python
job_embedding = model.encode([job_description])
resume_embeddings = model.encode(resumes)
```

```python
similarity_scores = cosine_similarity(job_embedding, resume_embeddings)[0]
similarity_scores
```

```python
results = pd.DataFrame({
    "Resume Index": range(len(resumes)),
    "Similarity Score": similarity_scores
})

results = results.sort_values(by="Similarity Score", ascending=False)
results
```

```python
for index, row in results.iterrows():
    print("Resume Index:", row["Resume Index"])
    print("Similarity Score:", round(row["Similarity Score"], 3))
    print("Resume Text:", resumes[int(row["Resume Index"])])
    print("-" * 60)
```

- This notebook implements a multi-resume ranking system using sentence embeddings.
- The job description and multiple resumes are converted into vector representations.
- Cosine similarity is computed to score each resume against the job description.
- Resumes are ranked based on similarity score to identify the best match.
- This approach can be extended to real-world hiring and recruitment systems.

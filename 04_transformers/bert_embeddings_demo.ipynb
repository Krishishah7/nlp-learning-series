```python
import torch
from transformers import AutoTokenizer, AutoModel
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")
```

```python
sentences = [
    "Machine learning is transforming technology",
    "Artificial intelligence is changing the world",
    "I enjoy cooking food",
    "Natural language processing is a part of AI"
]
```

```python
inputs = tokenizer(
    sentences,
    return_tensors="pt",
    padding=True,
    truncation=True
)
```

```python
with torch.no_grad():
    outputs = model(**inputs)
```

```python
token_embeddings = outputs.last_hidden_state
token_embeddings.shape
```

```python
cls_embeddings = token_embeddings[:, 0, :]
cls_embeddings.shape
```

```python
attention_mask = inputs["attention_mask"]
mask = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()

masked_embeddings = token_embeddings * mask
mean_embeddings = masked_embeddings.sum(dim=1) / mask.sum(dim=1)

mean_embeddings.shape
```

```python
similarity_matrix = cosine_similarity(mean_embeddings.numpy())
similarity_matrix
```

```python
score = cosine_similarity(
    mean_embeddings[0].unsqueeze(0).numpy(),
    mean_embeddings[1].unsqueeze(0).numpy()
)

score
```

- This notebook demonstrates how to generate sentence embeddings using a pretrained BERT model.
- Both CLS token embeddings and mean pooling strategies are used to represent sentences.
- Cosine similarity is applied to compare sentence embeddings and measure semantic similarity.
- This helps understand how BERT embeddings can be used for downstream NLP tasks.
